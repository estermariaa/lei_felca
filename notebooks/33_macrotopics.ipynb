```python
import pandas as pd
```

```python
df_topicos = pd.read_csv("modelo_final_escolhido/df_com_topicos.csv")
df_dados = pd.read_csv("../data/preprocessed_shorts.csv")
display(df_topicos)
display(df_dados)
```

```python
print(df_dados.columns)
print(df_topicos.columns)
```

```python
mapa_macros = {
    # MACRO 1: Adultização e impactos
    **{i: 'Adultização e Impactos' for i in [0, 22, 37, 41]},
    # MACRO 2: Exploração e casos
    **{i: 'Exploração e Abuso' for i in [7, 29, 33, 35, 38]},
    # MACRO 3: Regulação e ECA digital
    **{i: 'Regulação e ECA Digital' for i in [1, 4, 5, 11, 26, 34, 36]},
    # MACRO 4: Privacidade e tecnologia
    **{i: 'Privacidade e Infraestrutura' for i in [17, 21, 30, 39]},
    # MACRO 5: Games e mercado
    **{i: 'Games e Mercado' for i in [2, 13, 15]},
    # MACRO 6: Roblox e mobilização
    **{i: 'Roblox e Mobilização' for i in [10, 31, 32, 45, 46, 48]},
    # MACRO 7: Felca como ator midiático
    **{i: 'Felca como Ator Midiático' for i in [9, 20, 23, 24, 44]},
    # MACRO 8: Polarização e censura
    **{i: 'Polarização e Guerra Cultural' for i in [25, 27, 28, 49]},
    # MACRO 9: Educação e prevenção
    **{i: 'Educação e Prevenção' for i in [16]},
    # MACRO 10: Ruído / Não-substantivo
    **{i: 'Ruído / Não-substantivo' for i in [3, 6, 8, 12, 14, 18, 19, 40, 42, 43, 47]}
}

df_topicos['macrotopico'] = df_topicos['Topico_Atribuido'].map(mapa_macros)
```

```python
df_completo = pd.merge(
    df_topicos, 
    df_dados, 
    left_on='id', 
    right_on='videoId', 
    how='left'
)

colunas_para_manter = [
    'videoId', 'title', 'text', 'clean_text', 'Topico_Atribuido', 'macrotopico', 
    'channelTitle', 'channelId', 'publishedAt', 'description', 'tags', 
    'categoryId', 'duration', 'caption', 'viewCount', 'likeCount', 
    'commentCount', 'algorithmTopics', 'madeForKids', 'defaultAudioLanguage', 'isShorts'
]

df_final = df_completo[colunas_para_manter]
print(df_final.info())
display(df_final)
```

```python
tabela_formatos = pd.crosstab(
    df_final['macrotopico'], 
    df_final['isShorts'], 
    normalize='index' # Transforma em porcentagem
) * 100

print(tabela_formatos.round(1))
```

```python
engajamento = df_completo.groupby('macrotopico')[['viewCount', 'likeCount', 'commentCount']].median()
engajamento = engajamento.sort_values(by='viewCount', ascending=False)

print(engajamento)
```

```python
tabela_kids = pd.crosstab(df_completo['macrotopico'], df_completo['madeForKids'])
print(tabela_kids)
```
