# Análises das Caracteristicas Linguisticas Principais das Transcrições e Comentários por Macro-Tópico

```python
import pandas as pd
from pathlib import Path
```

```python
BASE_DIR = Path.cwd().parent 

LFTK_TRANSC = BASE_DIR / "data/lftk_transc.csv"
TOPICS_TRANSC = BASE_DIR / "data/df_macrotopics.csv"
MEAN_TOPICS_TRANSC = BASE_DIR / "data/lftk_analysis_transc.csv"
```

```python
def lftk_analysis(df: pd.DataFrame, features: list, output: str):
    for col in features:
        df[col] = pd.to_numeric(df[col], errors='coerce')

    medias_por_topico = df.groupby('macrotopico')[features].mean()

    tabela_medias = medias_por_topico.round(3)

    tabela_transposta = tabela_medias.T

    print(tabela_transposta)

    print("\n--- Código LaTeX ---")
    print(tabela_transposta.to_latex())

    tabela_transposta.to_csv(output)
```

## Transcrições

```python
df_completo_transc = pd.read_csv(LFTK_TRANSC, low_memory=False) 
df_topicos = pd.read_csv(TOPICS_TRANSC, low_memory=False)

"""df_completo_transc = pd.merge(
    df_lftk_transc, 
    df_topicos[['videoId', 'macrotopico', 'clean_text', 'text']],
    on='videoId', 
    how='inner'
)"""

top_10_features_transc = [
    'cole',
    'a_adp_pw',
    'a_cconj_ps',
    't_syll3',
    'a_punct_pw',
    'a_subtlex_us_zipf_pw',
    'a_pron_pw',
    'a_bry_ps',
    'uber_ttr',
    'a_adj_pw'
]
```

```python
lftk_analysis(df_completo_transc, top_10_features_transc, MEAN_TOPICS_TRANSC)
```
